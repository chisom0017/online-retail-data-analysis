```python
top_products = df. groupby("Description")["Quantity"].sum().sort_values(ascending = False)

top_products.head(10)
```

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
```

```python
top_countries = df.groupby("Country")["Revenue"].sum().sort_values(ascending = False)

top_countries.head(10)
```

```python
df["Month"] = df["InvoiceDate"].dt.to_period("M")
```

```python
df.head()
```

```python
monthly_sales = df.groupby("Month")["Revenue"].sum()
monthly_sales
```

```python
customer_sales = df.groupby("CustomerID")["Revenue"].sum().sort_values(ascending= False)

customer_sales.head(10)
```

```python
most_purchased = df.groupby("Description")["InvoiceNo"].count().sort_values(ascending = False)

most_purchased.head(10)
```

```python
import matplotlib.pyplot as plt

top_products.head(10).plot(kind="bar", figsize=(10,5))

plt.title("Top 10 Selling Products")
plt.xlabel("Product")
plt.ylabel("Quantity sold")
plt.xticks(rotation=90)
plt.show()
```

```python
top_countries.head(10).plot(kind="bar", figsize=(10,5))

plt.title("Top 10 Revenue-Genarating Countries")
plt.xlabel("Country")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
monthly_sales.plot(kind= "line", figsize=(10,5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)

plt.show()
```

```python
customer_sales.head(10).plot(kind="bar", figsize=(10,5))

plt.title("Top 10 Customers by Revenue")
plt.xlabel("Customerr ID")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
most_purchased.head(10).plot(kind="bar", figsize=(10,5))

plt.title("Top 10 Most Purchased Products")
plt.xlabel("Product")
plt.ylabel("Number of Purchases")
plt.xticks(rotation=90)
plt.show()
```

# Top Insights
Insight 1: The United Kingdom is the Largest Revenue Contributor

The analysis showed that the United Kingdom generated the highest revenue among all countries. This indicates that the UK is the company's strongest market and contributes the largest share of total sales.

Supported by: Top Revenue-Generating Countries chart.

Insight 2: Sales Reached Their Peak in November 2011

The monthly sales trend revealed that November 2011 recorded the highest revenue. This suggests increased customer purchasing during the holiday shopping season or promotional periods.

Supported by: Monthly Sales Trend line chart.

Insight 3: Customer 14646 Is the Highest-Value Customer

Customer 14646 generated the highest total revenue in the dataset. This highlights the importance of identifying and retaining high-value customers through loyalty programs and personalized offers.

Supported by: Top Customers by Revenue bar chart.

Insight 4: Best-Selling and Most Frequently Purchased Products Are Different

The analysis showed that WORLD WAR 2 GLIDERS ASSTD DESIGNS had the highest quantity sold, while WHITE HANGING HEART T-LIGHT HOLDER appeared in the highest number of purchases. This demonstrates that a product can be ordered frequently without having the highest total quantity sold.

Supported by: Top Selling Products and Most Purchased Products charts.

Insight 5: Data Cleaning Improved Analysis Accuracy

Removing 5,268 duplicate records and handling missing values improved the quality of the dataset and ensured that the analysis and visualizations were based on accurate data.

Supported by: Data Cleaning process.
